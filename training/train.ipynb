```python
data = pd.read_csv(r"C:\Users\Lenovo\OneDrive\Desktop\Kiln_test\data\earth.csv")
data.head()
```

```python
## DROPPING THE DATA COLS ROW NUMBER CUSTOMERID AND SURNAME
data = data.drop(['cycle_id', 'o2_percent'], axis=1)
data
```

```python
#OHE= GEO COL
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()
ohe_cb = ohe.fit_transform(data[['clay_body']])
ohe_cb
```

```python
ohe_cb.toarray()
```

```python
ohe.get_feature_names_out(["clay_body"])
```

```python
cb_encoded_df = pd.DataFrame(ohe_cb.toarray(), columns= ohe.get_feature_names_out(["clay_body"]))
cb_encoded_df
```

```python
data = pd.concat([data.drop('clay_body', axis=1), cb_encoded_df],axis=1)
data
```

```python
with open('ohe.pkl', 'wb') as file:
    pickle.dump(ohe, file)
```
